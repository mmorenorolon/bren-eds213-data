# Querying Data with Duckdb in Python

Steps for querying a database:

- Create the connection
- Create the cursor
- Formulate an SQL statement as a string
- Give the SQL statement to the cursor you created
- Fetch the results

## Import libraries

```python
import pandas as pd
import duckdb
```

## Open a Connection

```python
# Open connection to the database
conn = duckdb.connect('database/database.duckdb')
```

## Queries can be executed directly to the database connection

```python
# Obtain the data with fetchall
conn.sql("SELECT * FROM Site LIMIT 5").fetchall()
```

```python
# Return only one row
conn.sql("SELECT * FROM Site LIMIT 5").fetchone()
```

```python
# Returns the query as a dataframe 
conn.sql("SELECT * FROM Site LIMIT 5").df()
```

## Add a Cursor

```python
# Use a cursor
cur = conn.cursor()
cur.execute("SELECT * FROM Site")
```

```python
# Apply fetchall and all previous functions directly on the cursor
cur.fetchall()
# Iterate row by row
# Note: with an error, the cursor will 'disappear' and the cursor will have to be run again
# Metaphor: think of it like a tunnel that has not fully executed the query but it has stored it.
# Each cursor maintains the context of the pipelines (cursors) that can be executed 
```

```python
cur.execute("SELECT * FROM Site")
cur.fetchone()
```

```python
cur.fetchone()[0]
```

```python
# Iterate through the rows and print the site code
cur.execute("SELECT * FROM Site")

for row in cur.fetchall():
    print(row[0])
```

```python
# Iterations can be used for the cursor
cur.execute("SELECT * FROM Site")

while True:
    row = cur.fetchone()
    if row == None:
        break # What does break execute? It exits the loop
    print(f"Site code: {row[0]}")
```

Fetch all is slow especially if tables are big. Cursors are faster for working with large databases because you are only extracting only the rows that are needed as opposed to entire tables. 

```python
cur.execute("""
            CREATE TEMP TABLE Temp_nests AS
            SELECT * FROM Bird_nests
            """)
```

```python
cur.execute("SELECT * FROM Temp_nests LIMIT 3").fetchall()
```

## How to Close Your Connection

```python
# Close your connection
conn.close()
```

Tip: Don't share your connection (duckdb limitation). Duckdb does not support concurrency in the databases.
